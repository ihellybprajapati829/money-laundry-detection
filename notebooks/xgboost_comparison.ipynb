```python
import numpy as np, json
from xgboost import XGBClassifier
from sklearn.metrics import precision_recall_fscore_support, average_precision_score

d = np.load("../results/split_1_24.npz")
Xn, yn, tr, te = d["X"], d["y"], d["tr"], d["te"]

ratio = (yn[tr] == 0).sum() / (yn[tr] == 1).sum()
xgb = XGBClassifier(n_estimators=100, max_depth=6, random_state=42,
                    eval_metric="logloss", scale_pos_weight=ratio)
xgb.fit(Xn[tr], yn[tr])

pred = xgb.predict(Xn[te])
prob = xgb.predict_proba(Xn[te])[:, 1]
p, r, f1, _ = precision_recall_fscore_support(yn[te], pred, pos_label=1, average="binary")
print("XGBoost (train 1-24, weighted): P", round(p, 3), "R", round(r, 3), "F1", round(f1, 3),
      "PR-AUC", round(average_precision_score(yn[te], prob), 3))
```
